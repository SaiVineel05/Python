```python
names = ["raM", "RAJ", "Rose", "JaCk", "PaVi"]
marks = [[30,40,50],[66,77,55],[33,78,90],[87,97,57],[47,88,44]]

for i in range(5):
    avg = sum(marks[i]) / 3

    if avg > 80:
        g = "S"
    elif avg < 40:
        g = "F"
    elif 40 <= avg < 60:
        g = "B"
    else:
        g = "A"

    print("Mr/MS {} has scored {:.2f}% - {}".format(names[i], avg, g))

```

```python
import random

a = int(input("Enter num of teams: "))
teams = []
print("Enter all teams names:")
for i in range(a):
    t = input(f"Enter team {i+1} name: ")
    teams.append(t)

m = int(input("Enter num of meets between teams: "))
matches = []

for i in range(0, a - 1):
    for j in range(i + 1, a):
        for k in range(m):
            matches.append([teams[i], teams[j]])

random.shuffle(matches)

pos = 1
for match in matches:
    print("Match {} : {} vs {}".format(pos, match[0], match[1]))
    pos = pos + 1
```
