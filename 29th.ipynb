```python
a=input("Enter the password: ")
up=0
sm=0
sp=0
d=0
if len(a)>7:
  for i in a:
    if i.isupper():
      up=up+1
    elif i.islower():
      sm=sm+1
    elif i.isdigit():
      d=d+1
    else:
      sp=sp+1
  if up>0 and sm>0 and d>0 and sp>0:
      print("Strong")
  else:
      print("Weak")
else:
      print("It consists less characters")
```
