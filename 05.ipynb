```python
def karthikeya(a):
    if a==1:
       return
    i=2
    while(a%i !=0):
         i=i+1
    print(i,end=" ")
    karthikeya(a/i)

a=int(input("enter the number"))
karthikeya(a)
```

```python
class Rect:
  def __init__ (self,a,b):
       self.area=a*b
  def display(self):
       print("Area is ",self.area)

r=Rect(20,30)
r2=Rect(10,20)
r.display()
r2.display()
```
