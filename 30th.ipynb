```python
names = ["raM", "RAJ", "Rose", "JaCk", "PaVi"]
marks = [[30,40,50],[66,77,55],[33,78,90],[87,97,57],[47,88,44]]

for i in range(5):
    avg = sum(marks[i]) / 3

    if avg > 80:
        g = "S"
    elif avg < 40:
        g = "F"
    elif 40 <= avg < 60:
        g = "B"
    else:
        g = "A"

    print("Mr/MS {} has scored {:.2f}% - {}".format(names[i], avg, g))

```
